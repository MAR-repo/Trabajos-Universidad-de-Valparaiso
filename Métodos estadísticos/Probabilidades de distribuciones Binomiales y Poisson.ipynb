```python
def Binomial(p,n):
    if not float(n).is_integer() or n < 0:
        return print('El parámetro "n" debe ser entero positivo o cero')
    if p < 0 or p > 1:
        return print('El parámetro "p" debe pertenecer al intervalo ]0,1[')
    q = 1 - p
    B = q**n
    Bin = []
    Bin.append(B)
    for k in range(0,int(n)):
        B_k = B*p*(n-k)/(q*(k+1))
        B = B_k
        Bin.append(B)
    return (Bin)
```

```python
def Poisson(x,n):
    if x <= 0:
        return print("La media ingresada debe ser mayor a cero")
    if not float(n).is_integer() or n < 0:
        return print('El parámetro "n" debe ser entero positivo o cero')
    P = np.e**(-x)
    Pois = []
    Pois.append(P)
    for k in range(0,int(n)):
        P_k = P*x/(k+1)
        P = P_k
        Pois.append(P)
    return Pois
```

```python
def CDF(L,x,comp="menor"):
    if not float(x).is_integer() or x < 0:
        return print('El parámetro "x" debe ser entero positivo o cero')
    if x > len(L):
        return print('El parámetro "x" debe ser igual o menor a la cantidad de elementos de la lista')
    prob = 0
    for k in range(0,x+1):
        prob += L[k]
    if comp == "mayor":
        prob = 1 - prob
    return prob
```

```python
P45 = Poisson(45,100)
a = CDF(P45,50)
print(P45)
print(a)
```

```python
#Binomiales
#a) P(140 ≤ X ≤ 200), X ∼ Bin(1000, 1/6).

a = CDF(Binomial(1/6,1000),200) -  CDF(Binomial(1/6,1000),140) + Binomial(1/6,1000)[140]
print(a)
```

```python
#b) P(X ≥ 350), X ∼ Bin(500, 1/4).
b = CDF(Binomial(1/4,500),350,"mayor") + Binomial(1/4,500)[350]
print(sum(Binomial(1/4,500)))
print(b)
```

```python
#Poisson
#c)  P(X ≥ 50), X ∼ Poi(55).
c = CDF(Poisson(55,500),50,"mayor") + Poisson(55,500)[50]
print(c)
```
